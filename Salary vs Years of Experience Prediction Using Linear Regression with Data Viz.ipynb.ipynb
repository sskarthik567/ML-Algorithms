```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
```

```python
data = df[["YearsExperience","Salary"]].dropna()
data.shape
```

```python
plt.figure()
plt.scatter(data["YearsExperience"],data["Salary"])
plt.xlabel("YearsExperience")
plt.ylabel("Salary")
plt.title("YearsExperience vs Salary")
plt.show()
```

```python
x=data[["YearsExperience"]]
y=data["Salary"]

model  = LinearRegression()

model.fit(x,y)
```

```python
predicted_salary = model.predict(x)

plt.figure()
plt.scatter(x,y)
plt.plot(x, predicted_salary)
plt.xlabel("YearsExperience")
plt.ylabel("Salary")
plt.title("Best Fit Line")
plt.show()
```

```python
slope = model.coef_[0]
print("slope:",slope)

intercept = model.intercept_
print("intercept",intercept)
```

```python
mse = mean_squared_error(y, predicted_salary)
print("Mean Squares Error (MSE):",mse)
```

```python
example_prediction = model.predict([[7]])
print("Predicted Salary for 7 years:", example_prediction[0])
```

```python
rmse= np.sqrt(mse)
print("Root Mean Squared Error (RMSE):", rmse)

r2 = r2_score(y,predicted_salary)
print("R-Squared (R 2\u00B2):", r2)
```

```python
residuals = y - predicted_salary

plt.figure()
plt.scatter(predicted_salary,residuals)
plt.axhline(y=0)
plt.xlabel("Predicted Salary")
plt.ylabel("Residuals")
plt.title("Residual Plot")
plt.show()
```
